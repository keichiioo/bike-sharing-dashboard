## Menentukan Pertanyaan Bisnis

**SMART Question** adalah sebuah framework untuk merumuskan pertanyaan secara terstruktur agar memeperoleh informasi yang mendalam.

* **Spesific (Spesifik)**
  * Pertanyaan harus jelas, fokus pada sebuah topik tertentu, dan tidak bermakna ganda. Hindari pertanyaan yang terlalu luas.
    * Salah: Bagaimana cara meningkatkan penjualan?
    * Benar: Faktor apa saja yang memengaruhi penurunan penjualan produk kategori elektronik di wilayah Jakarta selama kuartal terakhir?
* **Measurable (Terukur)**
  * Pertanyaan harus bisa dijawab dengan angka atau matrix yang konkret, harus tahu apa yang akan dihitung.
    * Salah: Apakah pelanggan senang dengan layanan kita?
    * Benar: Berapa skor rata-rata Customer Satisfaction untuk layanan purna jual bulan ini dibandingkan bulan lalu?
* **Action-Oriented (Berorientasi Aksi)**
  * Hasil dari pertanyaan harus bisa memberikan arahan untuk melakukan tindakan nyata. Jika pertanyaan terjawab, stakeholder harus tahu apa langkah selanjutnya.
    * Salah: Mengapa orang suka berbelanja?
    * Benar: Fitur apa pada aplikasi yang paling sering digunakan sebelum pengguna memutuskan untuk melakukan checkout?
* **Relevant (Relevan)**
  * Hasil dari pertanyaan harus sejalan dengan tujuan utama bisnis atau masalah yang sedang dihadapi.
    * Salah: Menanyakan tentang stok gudang saat masalah utamanya adalah efektivitas kampanye media sosial.
    * Benar: Apakah kampanye iklan di Instagram memberikan Return on Ad Spend (ROAS) yang lebih tinggi dibandingkan iklan di TikTok?
* **Time-bound (Terikat Waktu)**
  * Pertanyaan harus ada batasan waktu yang jelas agar analisis memiliki konteks yang tepat.
    * Salah: Berapa banyak pengguna baru kita?
    * Benar: Berapa tingkat pertumbuhan pengguna baru secara bulanan (Month-over-Month) sepanjang tahun 2025?

**Contoh pertanyaan bisnis yang memenuhi seluruh elemen SMART**

***"Faktor apa saja yang memengaruhi penurunan conversion rate pada pengguna aplikasi Android di wilayah Jabodetabek sebesar 5% selama periode Flash Sale Maret 2026?"***

Keterangan:

- **Specific**: Fokus pada "penurunan conversion rate" untuk "aplikasi Android" di "Jabodetabek". Bukan sekadar penjualan turun.
- **Measurable**: Ada angka konkret yang ingin dianalisis, yaitu penurunan sebesar "5%".
- **Action-Oriented**: Dengan mengetahui faktor penyebabnya misalnya bug pada versi Android tertentu atau kendala logistik di Jabodetabek, tim bisa langsung melakukan perbaikan teknis atau operasional.
- **Relevant**: Penurunan konversi saat Flash Sale adalah masalah kritis bagi bisnis retail/e-commerce.
- **Time-bound**: Dibatasi pada periode spesifik "Flash Sale Maret 2026".

- **Pertanyaan 1:** Pada hari apa saja dalam seminggu (weekday) rata-rata total peminjaman sepeda (cnt) berada di bawah rata-rata harian tahunan sebesar lebih dari 5%, serta faktor cuaca atau status hari libur apa yang memengaruhinya selama tahun 2012?

### Cleaning Data

#### Fixing format problem

```python
hour_df['dteday'] = pd.to_datetime(hour_df['dteday'])
day_df['dteday'] = pd.to_datetime(day_df['dteday'])
```

#### Fixing data value problem

```python
def value_conversion(data):
  data['season_conv'] = data['season'].map({1: 'springer',
                                            2: 'summer',
                                            3: 'fall',
                                            4: 'winter'})
  data['yr_conv'] = data['yr'].map({0: 2011,
                                        1: 2012})
  data['holiday_conv'] = data['holiday'].map({1: 'yes',
                                              0: 'no'})
  data['workingday_conv'] = data['workingday'].map({1: 'yes',
                                                    0: 'no'})
  data['weekday_conv'] = data['weekday'].map({0: 'sunday',
                                              1: 'monday',
                                              2: 'tuesday',
                                              3: 'wednesday',
                                              4: 'thursday',
                                              5: 'friday',
                                              6: 'saturday'})
  data['temp_conv'] = data['temp']*41
  data['atemp_conv'] = data['atemp']*50
  data['hum_conv'] = data['hum']*100
  data['windspeed_conv'] = data['windspeed']*67
```

```python
value_conversion(hour_df)
value_conversion(day_df)

display(hour_df.head())
display(day_df.head())
```

**Insight:** (Opsional)
- Data sudah berhasil dibersihkan
- Ada kemungkinan hari bukan hari libur namun libur karena libur nasional. Jenis hari seperti ini harus diinvestigasi

## Visualization & Explanatory Analysis

### Pertanyaan 1: Pada hari apa saja dalam seminggu (weekday) rata-rata total peminjaman sepeda (cnt) berada di bawah rata-rata harian tahunan sebesar lebih dari 5%, serta faktor cuaca atau status hari libur apa yang memengaruhinya selama tahun 2012?

#### Menghitung Rata-Rata Peminjaman Harian pada 2012

```python
mean_cnt_2012 = day_df[(day_df['yr']==1)]['cnt'].mean()
print(f'rata_rata pinjaman sepeda tahun 2012: ~ {round(mean_cnt_2012)} sepeda/hari')
```

#### Investigasi pada hari apa penurunan terjadi

```python
growth_byday = (((day_df[day_df['yr']==1].groupby('weekday_conv')['cnt'].mean()-mean_cnt_2012)/mean_cnt_2012)*100).round(2).sort_values(ascending=True).reset_index(name='growth%')
growth_byday
```

```python
fig, ax = plt.subplots(figsize=(10,6))
colors = ['green' if val > -5 else 'red' for val in growth_byday['growth%']]
bars = ax.bar(growth_byday['weekday_conv'], growth_byday['growth%'], color=colors)
ax.bar_label(bars, fmt='%.2f%%', padding=3)
plt.axhline(y=-5, color='grey', linestyle='--', label='<5%')

plt.title('Pertumbuhan rata-rata pinjaman sepeda per hari terhadap rata-rata tahunan pada tahun 2012')
plt.xlabel('Hari')
plt.ylabel('Pertumbuhan rata-rata peminjaman sepeda (%)')
plt.legend()
plt.show()
```

### Pertanyaan 2: Pada 3 rentang jam teratas mana saja penyewaan pengguna terdaftar (registered) menyumbang lebih dari 50% total transaksi harian selama hari kerja di tahun 2012?

```python
total_reg_perhour = hour_df[(hour_df['workingday']==1) & (hour_df['yr']==1)].groupby('hr')['registered'].sum().sort_values(ascending=False).reset_index(name='total_registered')
total_reg_perhour.head()
```

```python
fig, ax = plt.subplots(figsize=(10, 7))

total_reg_sorted = total_reg_perhour.sort_values(by='total_registered', ascending=False)
y_labels = total_reg_sorted['hr'].astype(str)

bars = ax.barh(y_labels, total_reg_sorted['total_registered'], color='blue')
ax.bar_label(bars, padding=3, rotation=0, fontsize=8)

ax.invert_yaxis()

plt.ylabel('Jam (hr)')
plt.xlabel('Total peminjaman terdaftar (registered)')
plt.title('Total penyewaan terdaftar per jam pada hari kerja tahun 2012')
plt.tight_layout()
plt.show()
```

**Insight:** (Opsional)
- Pukul 5 sore, 8 pagi, dan 6 sore adalah 3 jam dengan peminjaman sepeda terdaftar paling tinggi.
- Jam dengan peminjaman sepeda terdaftar terbanyak disusul oleh pukul 7 malam dan 7 pagi.
